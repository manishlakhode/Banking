```python
import numpy as np
import matplotlib.pyplot as plt
import scipy.sparse as sp
from scipy.sparse.linalg import spsolve
from scipy.special import roots_legendre, roots_jacobi, eval_legendre
import time
from dataclasses import dataclass
from typing import Dict, List, Tuple, Optional, Callable
import seaborn as sns
from enum import Enum

# ============================================================================
# Enums and Configuration
# ============================================================================

class MethodType(Enum):
    CG = "Continuous Galerkin"
    DG = "Discontinuous Galerkin"

class IntegrationType(Enum):
    EXACT = "exact"
    INEXACT = "inexact"

# ============================================================================
# Configuration Classes
# ============================================================================

@dataclass
class SimulationConfig:
    """Configuration parameters for simulation."""
    u_vel: float = 2.0
    t_end: float = 1.0
    domain: Tuple[float, float] = (-1.0, 1.0)
    CFL: float = 0.25
    CFL_DG: float = 0.5  # DG requires smaller CFL
    sigma: float = 1.0 / 8.0

@dataclass
class SimulationResult:
    """Container for simulation results."""
    x: np.ndarray
    q_numerical: np.ndarray
    q_exact: np.ndarray
    L2_error: float
    wall_time: float
    Np: int
    N: int
    Ne: int
    method: MethodType
    integration_type: IntegrationType

@dataclass
class ConvergenceStudyConfig:
    """Configuration for convergence study."""
    # N_orders for convergence plotting (1, 4, 8, 16)
    N_orders: List[int] = None
    Ne_counts_cg: Dict[int, List[int]] = None
    Ne_counts_dg: Dict[int, List[int]] = None

    def __post_init__(self):
        # Using N=1, 4, 8, 16 as requested
        self.N_orders = [1, 4, 8, 16]

        # Defining the exact Ne counts required for each N:
        # N=1: Ne=16, 32, 64
        # N=4: Ne=4, 8, 16
        # N=8: Ne=2, 4, 8
        # N=16: Ne=1, 2, 4

        self.Ne_counts_cg = {
            1: [16, 32, 64],
            4: [4, 8, 16],
            8: [2, 4, 8],
            16: [1, 2, 4]
        }
        self.Ne_counts_dg = self.Ne_counts_cg.copy() # DG uses the same Ne counts

# ============================================================================
# Styling Configuration
# ============================================================================

class UnifiedPlotStyle:
    """Modern styling configuration for both CG and DG plots."""

    # Professional color palette
    COLORS = {
        'cg_numerical_exact': '#E74C3C',    # Vibrant red for CG Exact
        'cg_numerical_inexact': '#9B59B6',  # Purple for CG Inexact
        'dg_numerical_exact': '#2980B9',    # Blue for DG Exact
        'dg_numerical_inexact': '#2ECC71',  # Green for DG Inexact
        'exact_solution': '#2C3E50',        # Dark blue (Exact Ref)
        'element_boundaries': '#7F8C8D',
        'grid_lines': '#BDC3C7',
        'background': '#ECF0F1',
        'highlight': '#F39C12'
    }

    # Modern plot parameters
    PARAMS = {
        'figure.figsize': (12, 8),
        'font.size': 13,
        'font.family': 'DejaVu Sans',
        'axes.labelsize': 14,
        'axes.titlesize': 16,
        'legend.fontsize': 12,
        'xtick.labelsize': 12,
        'ytick.labelsize': 12,
        'grid.alpha': 0.4,
        'grid.linestyle': '-',
        'grid.linewidth': 0.5,
        'lines.linewidth': 2.5,
        'lines.markersize': 8
    }

    @classmethod
    def apply_style(cls, style: str = 'seaborn-v0_8-whitegrid'):
        """Apply modern plotting style."""
        plt.style.use(style)
        sns.set_palette("husl")
        plt.rcParams.update(cls.PARAMS)

# ============================================================================
# Core Numerical Methods Base Class (Unchanged)
# ============================================================================

class SpectralSolver:
    """Base class for spectral element methods."""

    def __init__(self, config: SimulationConfig):
        self.config = config

    def get_lgl_nodes_weights(self, N: int) -> Tuple[np.ndarray, np.ndarray]:
        """Calculate Legendre-Gauss-Lobatto nodes and weights."""
        if N == 1:
            return np.array([-1.0, 1.0]), np.array([1.0, 1.0])

        nodes_interior = roots_jacobi(N - 1, 1, 1)[0]
        nodes = np.concatenate(([-1.0], nodes_interior, [1.0]))

        P_N = eval_legendre(N, nodes)
        weights = 2.0 / (N * (N + 1) * P_N**2)

        return nodes, weights

    def lagrange_basis(self, N: int, nodes: np.ndarray, xi: np.ndarray) -> Tuple[np.ndarray, np.ndarray]:
        """Calculate Lagrange basis polynomials and derivatives."""
        Q = len(xi)
        psi = np.ones((Q, N + 1))
        dpsi = np.zeros((Q, N + 1))

        for i in range(N + 1):
            for j in range(N + 1):
                if i != j:
                    psi[:, i] *= (xi - nodes[j]) / (nodes[i] - nodes[j])

            for j in range(N + 1):
                if i != j:
                    term = 1.0 / (nodes[i] - nodes[j])
                    for k in range(N + 1):
                        if k != i and k != j:
                            term *= (xi - nodes[k]) / (nodes[i] - nodes[k])
                    dpsi[:, i] += term

        return psi, dpsi

    def _initial_condition(self, x: np.ndarray) -> np.ndarray:
        """Compute initial condition."""
        return np.exp(-(x / (2.0 * self.config.sigma))**2)

    def _time_integration(self, q0: np.ndarray, rhs_func: Callable, dt: float, Nt: int) -> np.ndarray:
        """Perform time integration using RK4."""
        q = q0.copy()
        for _ in range(Nt):
            k1 = rhs_func(q)
            k2 = rhs_func(q + 0.5 * dt * k1)
            k3 = rhs_func(q + 0.5 * dt * k2)
            k4 = rhs_func(q + dt * k3)
            q += (dt / 6.0) * (k1 + 2 * k2 + 2 * k3 + k4)
        return q

# ============================================================================
# CG and DG Solvers (Unchanged)
# ============================================================================

class CGSolver(SpectralSolver):
    """Continuous Galerkin Method solver."""

    def run_simulation(self, N: int, Ne: int, integration_type: IntegrationType) -> SimulationResult:
        """Run a single CG simulation."""
        t_start = time.time()

        # Grid setup (shared nodes)
        Np = N * Ne + 1
        lgl_nodes, lgl_weights = self.get_lgl_nodes_weights(N)
        x_global_nodes = self._construct_cg_grid(Ne, N, lgl_nodes)
        dx_element = (self.config.domain[1] - self.config.domain[0]) / Ne

        # Quadrature setup
        quad_nodes, quad_weights = self._setup_quadrature(integration_type, N, lgl_nodes, lgl_weights)
        psi_q, dpsi_q = self.lagrange_basis(N, lgl_nodes, quad_nodes)

        # Matrix assembly
        M_ref, D_ref = self._compute_reference_matrices(psi_q, dpsi_q, quad_weights)
        M_global, D_global = self._assemble_global_matrices(M_ref, D_ref, Ne, N, Np, dx_element)

        # Apply periodic boundary conditions
        M_global, D_global, x_eff = self._apply_periodic_bc(M_global, D_global, x_global_nodes, Np)

        # Initial condition and time integration
        q0 = self._initial_condition(x_eff)
        dt, Nt = self._compute_timestep_cg(dx_element, N)
        q = self._solve_cg_system(q0, M_global, D_global, integration_type, dt, Nt)

        return self._finalize_results(x_eff, q, t_start, N, Ne, MethodType.CG, integration_type, Np)

    def _construct_cg_grid(self, Ne: int, N: int, lgl_nodes: np.ndarray) -> np.ndarray:
        """Construct CG grid with shared nodes."""
        x_element_bnds = np.linspace(self.config.domain[0], self.config.domain[1], Ne + 1)
        Np = N * Ne + 1
        x_global_nodes = np.zeros(Np)

        for e in range(Ne):
            x0, x1 = x_element_bnds[e], x_element_bnds[e+1]
            J_map = (x1 - x0) / 2.0
            x_center = (x0 + x1) / 2.0
            x_nodes_in_element = x_center + lgl_nodes * J_map
            start_idx, end_idx = e * N, (e + 1) * N + 1
            x_global_nodes[start_idx:end_idx] = x_nodes_in_element

        return x_global_nodes[:Np - 1] # Return effective nodes after BC application for consistency

    def _setup_quadrature(self, integration_type: IntegrationType, N: int,
                         lgl_nodes: np.ndarray, lgl_weights: np.ndarray) -> Tuple[np.ndarray, np.ndarray]:
        if integration_type == IntegrationType.EXACT:
            return roots_legendre(N + 1)
        return lgl_nodes, lgl_weights

    def _compute_reference_matrices(self, psi_q: np.ndarray, dpsi_q: np.ndarray,
                                  quad_weights: np.ndarray) -> Tuple[np.ndarray, np.ndarray]:
        M_ref = np.einsum('ki,kj,k->ij', psi_q, psi_q, quad_weights)
        D_ref = np.einsum('ki,kj,k->ij', psi_q, dpsi_q, quad_weights)
        return M_ref, D_ref

    def _assemble_global_matrices(self, M_ref: np.ndarray, D_ref: np.ndarray,
                                Ne: int, N: int, Np: int, dx_element: float) -> Tuple[sp.csr_matrix, sp.csr_matrix]:
        M_global = sp.lil_matrix((Np, Np))
        D_global = sp.lil_matrix((Np, Np))
        J = dx_element / 2.0

        for e in range(Ne):
            global_indices = np.arange(e * N, (e + 1) * N + 1)
            M_global[np.ix_(global_indices, global_indices)] += M_ref * J
            D_global[np.ix_(global_indices, global_indices)] += D_ref

        return M_global.tocsr(), D_global.tocsr()

    def _apply_periodic_bc(self, M_global: sp.csr_matrix, D_global: sp.csr_matrix,
                          x_global_nodes: np.ndarray, Np: int) -> Tuple[sp.csr_matrix, sp.csr_matrix, np.ndarray]:
        M_global[0, :] += M_global[Np - 1, :]
        M_global[:, 0] += M_global[:, Np - 1]
        D_global[0, :] += D_global[Np - 1, :]
        D_global[:, 0] += D_global[:, Np - 1]

        Np_eff = Np - 1
        return (M_global[:Np_eff, :Np_eff].tocsr(),
                D_global[:Np_eff, :Np_eff].tocsr(),
                x_global_nodes[:Np_eff])

    def _compute_timestep_cg(self, dx_element: float, N: int) -> Tuple[float, int]:
        dx_min = dx_element / N
        dt = self.config.CFL * dx_min / self.config.u_vel
        Nt = int(np.ceil(self.config.t_end / dt))
        return self.config.t_end / Nt, Nt

    def _solve_cg_system(self, q0: np.ndarray, M_global: sp.csr_matrix, D_global: sp.csr_matrix,
                        integration_type: IntegrationType, dt: float, Nt: int) -> np.ndarray:
        if integration_type == IntegrationType.INEXACT:
            M_inv_diag = sp.diags(1.0 / M_global.diagonal())
            rhs_func = lambda q: -self.config.u_vel * (M_inv_diag @ (D_global @ q))
        else:
            rhs_func = lambda q: -self.config.u_vel * spsolve(M_global, D_global @ q)

        return self._time_integration(q0, rhs_func, dt, Nt)

    def _finalize_results(self, x: np.ndarray, q: np.ndarray, t_start: float,
                         N: int, Ne: int, method: MethodType, integration_type: IntegrationType,
                         Np: int) -> SimulationResult:
        wall_time = time.time() - t_start
        q_exact = self._initial_condition(x)

        # Calculate L2 error using the mass matrix for correct normalization (approximation here)
        L2_error = np.sqrt(np.sum((q - q_exact)**2) / np.sum(q_exact**2))

        # Recalculate Np based on effective degrees of freedom after BC (Np - 1)
        Np_eff = N * Ne

        return SimulationResult(x, q, q_exact, L2_error, wall_time, Np_eff, N, Ne, method, integration_type)


class DGSolver(SpectralSolver):
    """Discontinuous Galerkin Method solver."""

    def run_simulation(self, N: int, Ne: int, integration_type: IntegrationType) -> SimulationResult:
        """Run a single DG simulation."""
        t_start = time.time()

        # Grid setup (duplicated nodes)
        lgl_nodes, lgl_weights = self.get_lgl_nodes_weights(N)
        x_global_nodes = self._construct_dg_grid(Ne, N, lgl_nodes)
        dx_element = (self.config.domain[1] - self.config.domain[0]) / Ne
        Np = Ne * (N + 1)

        # Quadrature setup
        quad_nodes, quad_weights = self._setup_quadrature(integration_type, N, lgl_nodes, lgl_weights)
        psi_q, dpsi_q = self.lagrange_basis(N, lgl_nodes, quad_nodes)

        # Matrix assembly (block diagonal)
        M_ref, S_ref = self._compute_reference_matrices(psi_q, dpsi_q, quad_weights)
        M_global, S_global = self._assemble_global_matrices(M_ref, S_ref, Ne, N, Np, dx_element)

        # Initial condition and time integration
        q0 = self._initial_condition(x_global_nodes)
        dt, Nt = self._compute_timestep_dg(dx_element, N, lgl_nodes)
        q = self._solve_dg_system(q0, M_global, S_global, integration_type, Ne, N, dt, Nt)

        return self._finalize_results(x_global_nodes, q, t_start, N, Ne, MethodType.DG, integration_type, Np)

    def _construct_dg_grid(self, Ne: int, N: int, lgl_nodes: np.ndarray) -> np.ndarray:
        """Construct DG grid with duplicated nodes."""
        x_element_bnds = np.linspace(self.config.domain[0], self.config.domain[1], Ne + 1)
        Np = Ne * (N + 1)
        x_global_nodes = np.zeros(Np)

        for e in range(Ne):
            x0, x1 = x_element_bnds[e], x_element_bnds[e+1]
            J_map = (x1 - x0) / 2.0
            x_center = (x0 + x1) / 2.0
            x_nodes_in_element = x_center + lgl_nodes * J_map
            start_idx, end_idx = e * (N + 1), (e + 1) * (N + 1)
            x_global_nodes[start_idx:end_idx] = x_nodes_in_element

        return x_global_nodes

    def _setup_quadrature(self, integration_type: IntegrationType, N: int,
                         lgl_nodes: np.ndarray, lgl_weights: np.ndarray) -> Tuple[np.ndarray, np.ndarray]:
        if integration_type == IntegrationType.EXACT:
            return roots_legendre(N + 1)
        return lgl_nodes, lgl_weights

    def _compute_reference_matrices(self, psi_q: np.ndarray, dpsi_q: np.ndarray,
                                  quad_weights: np.ndarray) -> Tuple[np.ndarray, np.ndarray]:
        M_ref = np.einsum('ki,kj,k->ij', psi_q, psi_q, quad_weights)
        D_ref = np.einsum('ki,kj,k->ij', psi_q, dpsi_q, quad_weights)
        S_ref = D_ref.T
        return M_ref, S_ref

    def _assemble_global_matrices(self, M_ref: np.ndarray, S_ref: np.ndarray,
                                Ne: int, N: int, Np: int, dx_element: float) -> Tuple[sp.csr_matrix, sp.csr_matrix]:
        M_global = sp.lil_matrix((Np, Np))
        S_global = sp.lil_matrix((Np, Np))
        J = dx_element / 2.0

        for e in range(Ne):
            indices = np.arange(e * (N + 1), (e + 1) * (N + 1))
            M_global[np.ix_(indices, indices)] = M_ref * J
            S_global[np.ix_(indices, indices)] = S_ref

        return M_global.tocsr(), S_global.tocsr()

    def _compute_flux_terms(self, q_in: np.ndarray, Ne: int, N: int) -> np.ndarray:
        flux_term = np.zeros(Ne * (N + 1))

        for e in range(Ne):
            e_start_idx = e * (N + 1)
            e_end_node_idx = e * (N + 1) + N

            neighbor_e = Ne - 1 if e == 0 else e - 1
            neighbor_end_node_idx = neighbor_e * (N + 1) + N

            q_L_at_my_left_bnd = q_in[neighbor_end_node_idx]
            q_L_at_my_right_bnd = q_in[e_end_node_idx]

            f_star_left = self.config.u_vel * q_L_at_my_left_bnd
            f_star_right = self.config.u_vel * q_L_at_my_right_bnd

            flux_term[e_start_idx] = -f_star_left
            flux_term[e_end_node_idx] = f_star_right

        return flux_term

    def _compute_timestep_dg(self, dx_element: float, N: int, lgl_nodes: np.ndarray) -> Tuple[float, int]:
        dx_min_local = np.min(np.diff(lgl_nodes)) * (dx_element / 2.0)
        dt = (self.config.CFL_DG / (2 * N + 1)) * dx_min_local / self.config.u_vel
        Nt = int(np.ceil(self.config.t_end / dt))
        return self.config.t_end / Nt, Nt

    def _solve_dg_system(self, q0: np.ndarray, M_global: sp.csr_matrix, S_global: sp.csr_matrix,
                        integration_type: IntegrationType, Ne: int, N: int, dt: float, Nt: int) -> np.ndarray:
        def rhs_func(q_in):
            RHS = S_global @ (self.config.u_vel * q_in)
            flux_term = self._compute_flux_terms(q_in, Ne, N)
            return RHS - flux_term

        if integration_type == IntegrationType.INEXACT:
            M_inv = sp.diags(1.0 / M_global.diagonal())
            solve_func = lambda rhs: M_inv @ rhs
        else:
            solve_func = lambda rhs: spsolve(M_global, rhs)

        def wrapped_rhs_func(q):
            return solve_func(rhs_func(q))

        return self._time_integration(q0, wrapped_rhs_func, dt, Nt)

    def _finalize_results(self, x: np.ndarray, q: np.ndarray, t_start: float,
                         N: int, Ne: int, method: MethodType, integration_type: IntegrationType,
                         Np: int) -> SimulationResult:
        wall_time = time.time() - t_start
        q_exact = self._initial_condition(x)

        # Calculate L2 error using the mass matrix for correct normalization (approximation here)
        L2_error = np.sqrt(np.sum((q - q_exact)**2) / np.sum(q_exact**2))

        return SimulationResult(x, q, q_exact, L2_error, wall_time, Np, N, Ne, method, integration_type)

# ============================================================================
# Unified Visualization - THEORETICAL LINE REMOVED
# ============================================================================

class UnifiedVisualizer:
    """Modern visualization for both CG and DG results with 2 solution plots and 4 convergence plots."""

    def __init__(self):
        UnifiedPlotStyle.apply_style()

    def plot_two_solutions_combined(self, results: List[SimulationResult],
                                   save_path: Optional[str] = None):
        """Plot 2 combined plots: (CG Exact/Inexact) and (DG Exact/Inexact)."""

        cg_results = [r for r in results if r.method == MethodType.CG]
        dg_results = [r for r in results if r.method == MethodType.DG]

        fig, axes = plt.subplots(1, 2, figsize=(18, 8))

        plot_configs = [
            (cg_results, "Continuous Galerkin (CG) Method", axes[0], MethodType.CG),
            (dg_results, "Discontinuous Galerkin (DG) Method", axes[1], MethodType.DG)
        ]

        x_smooth = np.linspace(-1, 1, 1000)
        q_exact_smooth = np.exp(-(x_smooth / (2.0 * 0.125))**2)

        for results_list, title, ax, method in plot_configs:
            ax.plot(x_smooth, q_exact_smooth,
                   color=UnifiedPlotStyle.COLORS['exact_solution'],
                   linewidth=3, linestyle='--', alpha=0.8,
                   label='Exact Solution')

            for result in results_list:
                integration = result.integration_type
                label_prefix = f'{method.name}'

                if integration == IntegrationType.EXACT:
                    color = UnifiedPlotStyle.COLORS[f'{method.name.lower()}_numerical_exact']
                    label = f'{label_prefix} Exact (L²: {result.L2_error:.2e})'
                else:
                    color = UnifiedPlotStyle.COLORS[f'{method.name.lower()}_numerical_inexact']
                    label = f'{label_prefix} Inexact (L²: {result.L2_error:.2e})'

                if method == MethodType.CG:
                    ax.plot(result.x, result.q_numerical,
                           color=color, marker='o', markersize=4, linestyle='-',
                           linewidth=2, alpha=0.8, label=label)

                else:
                    for e in range(result.Ne):
                        start_idx = e * (result.N + 1)
                        end_idx = (e + 1) * (result.N + 1)
                        x_element = result.x[start_idx:end_idx]
                        q_element = result.q_numerical[start_idx:end_idx]

                        ax.plot(x_element, q_element,
                               color=color, marker='s', markersize=4, linestyle='-',
                               linewidth=2, alpha=0.8,
                               label=label if e == 0 else "")

            ax.set_title(title + f" (N={results_list[0].N}, Ne={results_list[0].Ne})",
                         fontsize=14, fontweight='bold', pad=15)
            ax.set_xlabel('Spatial Coordinate (x)', fontsize=12, fontweight='bold')
            ax.set_ylabel('Solution Amplitude (q)', fontsize=12, fontweight='bold')
            ax.grid(True, alpha=0.3)
            ax.legend(loc='upper right', framealpha=0.9, fontsize=10)
            ax.set_facecolor(UnifiedPlotStyle.COLORS['background'])

            if method == MethodType.DG and results_list:
                result = results_list[0]
                element_bnds = np.linspace(-1, 1, result.Ne + 1)
                for bnd in element_bnds:
                    ax.axvline(bnd, color=UnifiedPlotStyle.COLORS['element_boundaries'],
                              linestyle=':', alpha=0.6, linewidth=1)

        fig.suptitle('Comparison of Spectral Element Solutions: CG vs DG (Exact vs Inexact Integration)',
                    fontsize=16, fontweight='bold', y=0.98)

        plt.tight_layout(rect=[0, 0, 1, 0.96])

        if save_path:
            plt.savefig(save_path, dpi=300, bbox_inches='tight')

        plt.show()

    def plot_h_convergence_by_N(self, convergence_data: Dict[MethodType, Dict],
                                x_axis_key: str = 'Np',
                                save_path: Optional[str] = None):
        """
        Plots convergence for a FIXED polynomial order N, comparing all 4 methods (CG/DG, Exact/Inexact).
        Generates 4 plots in a 2x2 grid, one for each N in [1, 4, 8, 16].
        """
        N_ORDERS = [1, 4, 8, 16]

        fig, axes = plt.subplots(2, 2, figsize=(16, 12))
        axes = axes.flatten()

        method_styles = {
            (MethodType.CG, IntegrationType.EXACT): {'color': UnifiedPlotStyle.COLORS['cg_numerical_exact'], 'marker': 'o', 'label': 'CG Exact'},
            (MethodType.CG, IntegrationType.INEXACT): {'color': UnifiedPlotStyle.COLORS['cg_numerical_inexact'], 'marker': 'o', 'label': 'CG Inexact'},
            (MethodType.DG, IntegrationType.EXACT): {'color': UnifiedPlotStyle.COLORS['dg_numerical_exact'], 'marker': 's', 'label': 'DG Exact'},
            (MethodType.DG, IntegrationType.INEXACT): {'color': UnifiedPlotStyle.COLORS['dg_numerical_inexact'], 'marker': 's', 'label': 'DG Inexact'},
        }

        for i, N in enumerate(N_ORDERS):
            ax = axes[i]

            # --- CODE REMOVED ---
            # The section that calculated and plotted the theoretical convergence line is removed here.
            # --- END REMOVED CODE ---

            # Plot simulation data for all four methods for the current fixed N
            for (method, integration_type), style in method_styles.items():
                int_type_key = integration_type.value

                if N in convergence_data[method][int_type_key]:
                    data_list = convergence_data[method][int_type_key][N]
                    x_vals = [d[x_axis_key] for d in data_list]
                    y_vals = [d['L2_error'] for d in data_list]

                    ax.loglog(x_vals, y_vals,
                             color=style['color'], marker=style['marker'], markersize=8,
                             linestyle='-', linewidth=2.5, alpha=0.8,
                             label=style['label'])

            ax.set_title(f'Convergence for Polynomial Order N = {N}', fontsize=14, fontweight='bold', pad=15)
            ax.set_xlabel('Degrees of Freedom ($N_p$)', fontsize=12, fontweight='bold')
            ax.set_ylabel('Normalized L² Error', fontsize=12, fontweight='bold')
            ax.grid(True, alpha=0.3)
            ax.legend(loc='upper right', fontsize=10)
            ax.set_facecolor(UnifiedPlotStyle.COLORS['background'])

        fig.suptitle('Convergence Analysis: Methods Compared at Fixed Polynomial Order N (h-Refinement)',
                    fontsize=16, fontweight='bold', y=0.98)

        plt.tight_layout(rect=[0, 0, 1, 0.96])

        if save_path:
            plt.savefig(save_path, dpi=300, bbox_inches='tight')

        plt.show()

# ============================================================================
# Unified Analysis Workflow and Main Execution (Unchanged)
# ============================================================================

class UnifiedSpectralAnalysis:
    """Manager for unified CG and DG analysis."""

    def __init__(self, config: SimulationConfig, study_config: ConvergenceStudyConfig = None):
        self.config = config
        self.study_config = study_config or ConvergenceStudyConfig()
        self.cg_solver = CGSolver(config)
        self.dg_solver = DGSolver(config)
        self.visualizer = UnifiedVisualizer()

    def run_single_simulation(self, method: MethodType, N: int, Ne: int,
                            integration_type: IntegrationType) -> SimulationResult:
        """Run a single simulation with specified method."""
        if method == MethodType.CG:
            return self.cg_solver.run_simulation(N, Ne, integration_type)
        else:
            return self.dg_solver.run_simulation(N, Ne, integration_type)

    def run_four_case_study(self, N: int = 8, Ne: int = 8) -> List[SimulationResult]:
        """Run the 4 cases: CG exact, CG inexact, DG exact, DG inexact."""
        print("🔄 Running 4-Case Study: CG/DG with Exact/Inexact Integration")
        print("=" * 60)

        results = []
        cases = [
            (MethodType.CG, IntegrationType.EXACT, "CG with Exact Integration"),
            (MethodType.CG, IntegrationType.INEXACT, "CG with Inexact Integration"),
            (MethodType.DG, IntegrationType.EXACT, "DG with Exact Integration"),
            (MethodType.DG, IntegrationType.INEXACT, "DG with Inexact Integration")
        ]

        for method, integration_type, description in cases:
            print(f"Running {description} (N={N}, Ne={Ne})...")
            result = self.run_single_simulation(method, N, Ne, integration_type)
            results.append(result)
            print(f"  ✓ L² Error: {result.L2_error:.2e}")
            print(f"  ✓ Wall Time: {result.wall_time:.2f}s")
            print(f"  ✓ Degrees of Freedom: {result.Np}")
            print()

        return results

    def run_convergence_study(self) -> Dict[MethodType, Dict]:
        """Run comprehensive convergence study for both methods, only for the selected N orders and Ne counts."""
        print("📊 Running Unified Convergence Study")
        print("=" * 50)

        convergence_data = {
            MethodType.CG: {IntegrationType.EXACT.value: {}, IntegrationType.INEXACT.value: {}},
            MethodType.DG: {IntegrationType.EXACT.value: {}, IntegrationType.INEXACT.value: {}}
        }

        for method in [MethodType.CG, MethodType.DG]:
            print(f"\n🔍 Analyzing {method.value}...")

            Ne_counts = (self.study_config.Ne_counts_cg if method == MethodType.CG
                        else self.study_config.Ne_counts_dg)

            for integration_type in [IntegrationType.EXACT, IntegrationType.INEXACT]:
                print(f"  {integration_type.value} integration: ", end="")

                for N in self.study_config.N_orders:
                    convergence_data[method][integration_type.value][N] = []

                    if N in Ne_counts:
                        for Ne in Ne_counts[N]:
                            result = self.run_single_simulation(method, N, Ne, integration_type)

                            convergence_data[method][integration_type.value][N].append({
                                'Np': result.Np,
                                'L2_error': result.L2_error,
                                'wall_time': result.wall_time
                            })

                    print(f"N={N} ", end="")
                print()

        print("\n✅ Unified convergence study completed!")
        return convergence_data

# ============================================================================
# Main Execution
# ============================================================================

def main():
    """Main execution function for unified CG/DG analysis."""

    # Initialize configuration
    config = SimulationConfig(sigma=1.0/8.0)
    # Use the requested N and Ne counts
    study_config = ConvergenceStudyConfig()

    # Initialize analysis
    analysis = UnifiedSpectralAnalysis(config, study_config)

    print("🎯 UNIFIED SPECTRAL ELEMENT ANALYSIS (Theoretical slope removed from plots)")
    print("=" * 60)
    print("Four Case Study: CG/DG with Exact/Inexact Integration (N=8, Ne=8)")
    print("=" * 60)

    # Run the 4-case study
    print("\n1. RUNNING 4-CASE SOLUTION STUDY")
    four_case_results = analysis.run_four_case_study(N=8, Ne=8)

    # Plot the 2 combined solution plots
    print("\n📈 GENERATING 2 COMBINED SOLUTION PLOTS (CG & DG)")
    analysis.visualizer.plot_two_solutions_combined(four_case_results)

    # Run convergence study
    print(f"\n2. RUNNING CONVERGENCE ANALYSIS (N={study_config.N_orders})")
    convergence_data = analysis.run_convergence_study()

    # Plot 4 separate convergence plots, one for each N
    print("\n📊 GENERATING 4 CONVERGENCE PLOTS (N-specific h-Refinement, no theoretical line)")
    analysis.visualizer.plot_h_convergence_by_N(convergence_data)

    # Summary
    print("\n" + "=" * 60)
    print("📊 FOUR-CASE ANALYSIS SUMMARY")
    print("=" * 60)

    for result in four_case_results:
        method_name = result.method.value
        int_type = result.integration_type.value
        print(f"{method_name} ({int_type}):")
        print(f"  • L² Error: {result.L2_error:.2e}")
        print(f"  • Wall Time: {result.wall_time:.2f}s")
        print(f"  • DOF: {result.Np}")
        print()

    print("🎉 Analysis completed successfully with tailored convergence plots!")

if __name__ == "__main__":
    main()
```
